# Part 5 · Encapsulation: protect your data (35 min)

**Reminder**
| Name | Meaning |
|---|---|
| `name` | public: anyone may read and change it |
| `_name` | protected: "internal, please do not touch" (convention only) |
| `__name` | private: Python hides it from outside the class |

```python
class Product:
    def __init__(self, price):
        self.price = price            # goes through the setter below!

    @property
    def price(self):                  # getter: p.price
        return self.__price

    @price.setter
    def price(self, value):           # setter: p.price = 10
        if value < 0:
            raise ValueError("Negative price")
        self.__price = value

try:
    p = Product(-5)
except ValueError as e:
    print("Error:", e)
```
