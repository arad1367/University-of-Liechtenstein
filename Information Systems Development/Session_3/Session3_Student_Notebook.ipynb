## 1.1 The problem

Look at this. The same idea, written three times.

````python
print("Hello, Anna!")
print("Welcome back to MiniShop.")

print("Hello, Ben!")
print("Welcome back to MiniShop.")

print("Hello, Carla!")
print("Welcome back to MiniShop.")
````

Change the welcome text and you must change it in three places. Miss one and nobody notices.

## 1.2 The fix: name it once

````python
def greet(name):
    print(f"Hello, {name}!")
    print("Welcome back to MiniShop.")


greet("Anna")
greet("Ben")
greet("Carla")
````

Four parts to notice:

| Part | What it is |
|---|---|
| `def` | the keyword that starts a definition |
| `greet` | the name you choose — make it say what it does |
| `(name)` | the **parameter** it needs to do the job |
| the indented block | the body, which runs only when you call it |

## 1.3 Parameters and arguments

**Parameters** are the names in the definition. **Arguments** are the values you pass in.

````python
def apply_tax(amount, rate):
    return amount * (1 + rate)


print(apply_tax(120, 0.2))
print(apply_tax(rate=0.2, amount=120))   # named: order stops mattering
````

## 1.4 Default values

A parameter with a default becomes optional. Defaults go **last**, after the ones without.

````python
def apply_tax(amount, rate=0.077):      # 7.7% is Swiss VAT
    return round(amount * (1 + rate), 2)


print(apply_tax(100))          # 107.7  <- uses the default
print(apply_tax(100, 0.2))     # 120.0  <- overrides it
````

Why `round(..., 2)`? Without it the first line prints `107.69999999999999`.

That is the float problem from Session 2, and money is exactly where it bites.
Round at the edge, when you hand the number out.

## 1.5 `return` vs `print` — read this twice

This confuses everyone, so spend a minute here.

- `print` puts text **on the screen** for a human. Your program keeps nothing.
- `return` hands a value **back to your code**, so you can store it, test it, or use it again.

````python
def shows(x):
    print(x * 2)


def gives(x):
    return x * 2


a = shows(5)        # prints 10
b = gives(5)        # prints nothing

print(a)            # None  <- shows() handed back nothing
print(b)            # 10    <- gives() handed back a value
````

**A function with no `return` always returns `None`.**

This matters beyond style: a HackerRank function stub wants `return`, never `print`.
So does every Flask route you write later today.

A list remembers **order**, so you ask for `items[0]`.

## 2.1 Reading a field safely

`product["colour"]` raises a `KeyError` if the field is missing.
`product.get("colour")` returns `None` instead, and lets you choose a fallback.

````python
product = {"name": "Laptop", "price": 899.0}

print(product["name"])                 # Laptop
print(product.get("colour"))           # None
print(product.get("colour", "n/a"))    # n/a
````

Use square brackets when the field is **required** — you *want* the program to stop
if the data is wrong. Use `.get()` for optional fields.

## 2.2 Changing, adding, looping

The same line both changes an existing field and adds a new one. Python does not warn you.

````python
product = {"name": "Laptop", "stock": 4}

product["stock"] = 3             # changes it
product["colour"] = "silver"     # adds it

for field in product:            # looping gives you the KEYS
    print(field, "=", product[field])

print("---")

for field, value in product.items():    # .items() gives you both
    print(field, "=", value)
````

## 3.1 Asking the table a question

A function, a list, a dictionary, a loop and a condition — all in eight lines.
This is the shape of most real reporting code.

````python
def in_stock(rows):
    available = []
    for row in rows:
        if row["stock"] > 0:
            available.append(row["name"])
    return available


print(in_stock(catalogue))
````

Note that it **returns** the answer and lets the caller decide what to do with it.

# 5 · Flask — your script becomes a website

## Why this part needs VS Code

A Flask app is a **server**. It starts, then waits for a browser to ask it something,
and it keeps waiting until you stop it.

A notebook cell runs, finishes, and hands control back. Those two things do not fit
together — a cell that starts a server never finishes, so the notebook just hangs.

So from here on we **write real `.py` files** and run them from the terminal.
That is what the rest of this course looks like.

### How the web works, in three lines

1. **Request** — a browser asks for a URL.
2. **Server** — your Python decides what to send back.
3. **Response** — HTML arrives and the browser draws it.

Flask is a small library that lets an ordinary Python function answer step 2.

---

## 5.2 Write your first app

The `%%writefile` magic saves a cell to a file instead of running it.
This cell creates `app.py` next to your notebook.

````python
%%writefile app.py
from flask import Flask

app = Flask(__name__)


@app.route("/")
def home():
    return "Hello from MiniShop!"


if __name__ == "__main__":
    app.run(debug=True)
````

**To stop the server: `Ctrl + C` in the terminal.** You will need this constantly.

### What each piece does

| Line | What it does |
|---|---|
| `app = Flask(__name__)` | creates the application |
| `@app.route("/")` | says *this function answers that URL* |
| `return "Hello..."` | whatever you return is what the browser receives |
| `debug=True` | reloads on save and shows errors in the page |

Notice: the route function **returns**. It does not print. Exactly the lesson from Part 1.

## 5.3 More than one page

Each route is a URL plus a function. Angle brackets capture part of the URL and
hand it to the function as an argument.

````python
%%writefile app.py
from flask import Flask

app = Flask(__name__)


@app.route("/")
def home():
    return "Welcome to MiniShop"


@app.route("/about")
def about():
    return "A tiny shop built in Session 3"


@app.route("/product/<name>")
def product(name):
    return f"You asked for {name}"


if __name__ == "__main__":
    app.run(debug=True)
````

## 5.4 Real data, real HTML

Returning strings gets dull fast. **Templates** are HTML files with holes in them,
and Flask fills the holes with your data.

Two rules: the HTML lives in a folder called `templates`, and inside it
`{{ }}` prints a value while `{% %}` runs a loop or an `if`.

First the template — note the folder name in the writefile path.

````python
%%writefile templates/index.html
<!doctype html>
<html>
  <head>
    <title>MiniShop</title>
    <style>
      body { font-family: sans-serif; margin: 2rem; }
      li   { margin: 0.4rem 0; }
      .out { color: #999; }
    </style>
  </head>
  <body>
    <h1>MiniShop</h1>
    <p>{{ products|length }} products, {{ available }} in stock.</p>
    <ul>
      {% for p in products %}
        <li class="{% if p['stock'] == 0 %}out{% endif %}">
          {{ p["name"] }} — {{ p["price"] }} EUR
          {% if p["stock"] == 0 %}(sold out){% endif %}
        </li>
      {% endfor %}
    </ul>
  </body>
</html>
````

If that cell fails with *No such file or directory*, the `templates` folder does not
exist yet. Run the next cell first, then run the one above again.

````python
import os
os.makedirs("templates", exist_ok=True)
print("templates/ is ready")
````

Now the app. Look closely: `catalogue` is the **same list of dictionaries** from
Part 3, and `in_stock()` is the **same function**. Nothing new — it just has an
audience now.

````python
%%writefile app.py
from flask import Flask, render_template

app = Flask(__name__)

catalogue = [
    {"name": "Laptop", "price": 899.0, "stock": 4,  "kind": "computer"},
    {"name": "Mouse",  "price": 24.5,  "stock": 31, "kind": "accessory"},
    {"name": "Cable",  "price": 9.99,  "stock": 0,  "kind": "accessory"},
    {"name": "Hub",    "price": 149.0, "stock": 7,  "kind": "accessory"},
    {"name": "Screen", "price": 329.0, "stock": 2,  "kind": "computer"},
]


def in_stock(rows):
    available = []
    for row in rows:
        if row["stock"] > 0:
            available.append(row["name"])
    return available


@app.route("/")
def home():
    return render_template("index.html",
                           products=catalogue,
                           available=len(in_stock(catalogue)))


if __name__ == "__main__":
    app.run(debug=True)
````

`python app.py`, then reload the page. Your data is now a web page.

That is the whole trick: **a route returns the output of a function, and a template
turns that data into HTML.**
