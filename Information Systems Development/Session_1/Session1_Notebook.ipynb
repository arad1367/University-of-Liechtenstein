## 3 · Printing nicely with f-strings

The clumsy way needs `str()` and a lot of `+` signs:

```python
print("Price: " + str(price))
```

The f-string way is shorter and much easier to read. Put an `f` **before** the
opening quote, and put any value inside `{curly braces}`.

````python
print(f"Price: {price} EUR")
print(f"We have {stock} of the {name} in stock.")
````

## 4 · Asking the user

`input()` stops the program and waits. It hands back whatever the person typed.

````python
user_name = input("Your name: ")
print(f"Hello, {user_name}!")
````

### `input()` always gives you text

Even if the person types `30`, Python receives the **text** `"30"`, not the number 30.

Run the next cell and type a number. It will fail — that is the point.

````python
age = input("Your age: ")
age + 1        # this breaks on purpose
````

Read the error from the **bottom up**:

- the **last line** tells you the error type and the message
- the line above tells you **where** it happened

`TypeError` means you mixed text and numbers.

The fix is `int()`, which translates text into a whole number.

````python
age = int(input("Your age: "))
print(f"Next year you will be {age + 1}.")
````

## 6 · Division: `/` versus `//`

This distinction comes back in every later session.

````python
print(10 / 3)     # 3.3333333333333335  — always a float
print(10 // 3)    # 3                   — the whole part only
print(10 % 3)     # 1                   — what is left over
````

Business version: you have 10 items and boxes that hold 3.

- `10 // 3` → how many **full boxes** you can fill
- `10 % 3` → how many items are **left over**
