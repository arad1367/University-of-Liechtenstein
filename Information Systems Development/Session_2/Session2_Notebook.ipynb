# 1 · Conditions

## 1.1 Booleans

A **boolean** has only two possible values: `True` or `False`.
Capital T, capital F, and no quotes — they are not text.

````python
stock = 4

print(stock > 0)      # True
print(stock == 10)    # False
print(type(stock > 0))
````

## 1.2 The six comparison operators

| Operator | Reads as | Example | Result |
|---|---|---|---|
| `==` | is equal to | `4 == 4` | `True` |
| `!=` | is different from | `4 != 5` | `True` |
| `<`  | is smaller than | `3 < 5` | `True` |
| `>`  | is bigger than | `3 > 5` | `False` |
| `<=` | is smaller or equal | `5 <= 5` | `True` |
| `>=` | is bigger or equal | `6 >= 5` | `True` |

**`=` is not `==`.**
`stock = 4` is an instruction: put 4 into `stock`.
`stock == 4` is a question: is `stock` equal to 4?

````python
price = 899

print(price == 899)
print(price != 899)
print(price <= 900)
````

## 1.3 if / elif / else

Watch three things: the **colon** at the end of the line, the **four spaces** of indent,
and the fact that Python stops at the **first** branch that is `True`.

````python
stock = 10

if stock == 4:
    print("Sold out")
elif stock < 5:
    print("Only a few left")
else:
    print("In stock")
````

Change `stock` to `0`, then `3`, then `40`, and run the cell again each time.
Three different branches from the same code.

## 1.4 and, or, not

- `and` — **both** sides must be True. It narrows the group.
- `or` — **one** side is enough. It widens the group.
- `not` — flips True into False.

````python
total  = 120
member = True

if total >= 100 and member:
    discount = 0.20
elif total >= 100:
    discount = 0.10
else:
    discount = 0.0

print(f"With {discount} discount, You pay {total * (1 - discount)} EUR")
````

> **Careful.** Swap the first two branches and members silently lose their 20%.
> The code still runs. It just pays the wrong amount. That is the dangerous kind of bug.

# 2 · Repetition

## 2.1 while — repeat until something changes

````python
guesses = 0

while guesses < 3:
    print(f"Attempt {guesses + 1}")
    guesses = guesses + 1

print("Out of attempts")
````

````python
# crash topic: import ...
````

How it runs: check the condition → if True, run the indented block → go back to the
top → check again → if False, skip to the line after the loop.

### The infinite loop

```python
guesses = 0
while guesses < 3:
    print(guesses)      # nothing changes guesses — this never stops
```

**The habit that saves you:** before you run a `while` loop, point at the line that
will eventually make the condition `False`. If you cannot find it, it is not there.

To stop a runaway cell in Colab, press the square **Stop** button next to it.

## 2.2 for and range()

````python
products = ["Laptop", "Mouse", "Cable"]

for item in products:
    print(item)

print("---")

for n in range(3):
    print(n)          # 0, 1, 2
````

`range(3)` gives **three numbers starting at zero**: 0, 1, 2. It stops *before* 3.

- `range(5)` → 0 1 2 3 4
- `range(1, 6)` → 1 2 3 4 5
- `range(0, 10, 2)` → 0 2 4 6 8

**Which loop?** Can you count the turns before you start? If yes, `for`. If no, `while`.

## 2.3 The accumulator pattern

Start empty *outside* the loop, add one thing per turn *inside* the loop,
read the answer *after* the loop. This shape comes back in every remaining session.

````python
prices = [899.0, 24.5, 9.99]
total  = 0

for p in prices:
    total = total + p

print(f"Basket total: {total} EUR")
````

> Move `total = 0` inside the loop and run it again. The answer becomes wrong but the
> program does not crash. Silent wrong answers are the ones that reach production.

## 2.4 break and continue

````python
prices = [12, 45, 730, 8, 950]

for p in prices:
    if p > 500:
        print(f"First expensive item: {p}")
        break            # found it — leave the loop now

print("---")

for p in prices:
    if p == 0:
        continue         # skip this turn, go to the next item
    print(p)
````

# 3 · Lists

A list is one name holding many values. Square brackets make it, and square brackets
also pick items out of it.

````python
products = ["Laptop", "Mouse", "Cable", "Hub"]

print(products[0])       # Laptop     — counting starts at 0
print(products[-1])      # Hub        — negative counts from the end
print(products[0:2])     # ['Laptop', 'Mouse']  — up to, not including, 2
print(len(products))     # 4
````

| Position | 0 | 1 | 2 | -1 |
|---|---|---|---|---|
| `products[...]` | Laptop | Mouse | Cable | Hub |
| Reads as | first | second | third | last |

The last item is at index `len(products) - 1`. That is why `-1` is such a relief.

## 3.1 Growing a list

````python
basket = []                    # an empty list is a fine place to start

basket.append("Laptop")
basket.append("Mouse")

print(basket)
print(len(basket))

for item in basket:
    print(f"- {item}")
````

Note the dot: `basket.append(x)` — the list does it to itself.
`len(basket)` — something done *to* the list. You will meet many more dot-methods
in Session 4.

## 3.2 List + loop + condition

This is the whole session on one screen, and it is the shape of most real data work.

````python
prices = [899.0, 24.5, 9.99, 149.0]
cheap  = []

for p in prices:
    if p < 100:
        cheap.append(p)

print(f"{len(cheap)} products under 100 EUR")
print(cheap)
````

Read it aloud: *for every price in the list… if it is under a hundred… put it in the
cheap list.* Two levels of indent — count them.
